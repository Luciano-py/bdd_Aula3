```python
# Combinando condições com and / or / not dentro de um if
idade, tem_documento = 20, True
if idade >= 18 and tem_documento:
    print("Pode votar")

nota, faltas = 5.5, 18
if nota < 5 or faltas > 15:
    print("Reprovado")

chovendo = False
if not chovendo:
    print("Vou de bike")
```
